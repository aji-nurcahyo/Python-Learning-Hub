# Lab A03: Exploratory Data Analysis (EDA)
Evaluasi distribusi statistik, deteksi outlier IQR, dan matriks korelasi pada data `telecom_qos_kpi.csv`.

```python
from pathlib import Path
import pandas as pd

df = pd.read_csv(Path('datasets/telecom_qos_kpi.csv'))
q1, q3 = df['latency_ms'].quantile([0.25, 0.75])
iqr = q3 - q1
outliers = df[df['latency_ms'] > q3 + 1.5 * iqr]
assert 'TWR-09' in outliers['tower_id'].tolist()
print('Outliers:', outliers[['tower_id', 'latency_ms']].to_dict(orient='records'))
```
