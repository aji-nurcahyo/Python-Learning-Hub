# Lab B03: Unsupervised Learning & Dimensionality Reduction
Eksperimen K-Means, DBSCAN anomali, dan PCA pada `user_usage_profiles.csv`.

```python
from pathlib import Path
import pandas as pd
from sklearn.cluster import DBSCAN, KMeans
from sklearn.decomposition import PCA
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

df = pd.read_csv(Path('datasets/user_usage_profiles.csv'))
features = ['daily_gb', 'video_ratio', 'avg_latency_ms', 'active_minutes']
X_sc = StandardScaler().fit_transform(df[features])
labels = KMeans(n_clusters=3, n_init=10, random_state=42).fit_predict(X_sc)
print('Silhouette K=3:', round(float(silhouette_score(X_sc, labels)), 4))
pca = PCA(n_components=2).fit(X_sc)
print('PCA variance ratio:', [round(float(v), 4) for v in pca.explained_variance_ratio_])
```
