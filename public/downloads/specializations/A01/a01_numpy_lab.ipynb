# Lab A01: Numerical Computing with NumPy
Eksplorasi interaktif struktur `ndarray`, vectorization, broadcasting, dan analisis matriks sensor.

```python
from pathlib import Path
import numpy as np

csv_path = Path('datasets/sensor_matrix.csv')
data = np.genfromtxt(csv_path, delimiter=',', skip_header=1, dtype=np.float64)
assert data.shape == (12, 5)
print('Dataset shape:', data.shape)
```

```python
metrics = data[:, 1:]
mean = metrics.mean(axis=0, keepdims=True)
std = metrics.std(axis=0, keepdims=True)
z_scores = (metrics - mean) / np.where(std == 0, 1.0, std)
anomaly_rows = np.any(np.abs(z_scores) > 1.8, axis=1)
print('Node terdeteksi anomali:', data[anomaly_rows, 0].astype(int).tolist())
```
