# Lab A04: Data Visualization with Matplotlib & Seaborn
Pembuatan grafik statistik multi-panel OO API dari dataset `telecom_qos_kpi.csv`.

```python
import tempfile
from pathlib import Path
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv(Path('datasets/telecom_qos_kpi.csv'))
fig, ax = plt.subplots(1, 2, figsize=(10, 4), dpi=150)
sns.boxplot(data=df, x='region', y='latency_ms', ax=ax[0])
sns.heatmap(df[['traffic_gbps', 'latency_ms', 'packet_loss_pct']].corr(), annot=True, ax=ax[1])
out = Path(tempfile.gettempdir()) / 'a04_notebook_preview.png'
fig.tight_layout()
fig.savefig(out)
plt.close(fig)
assert out.exists()
```
