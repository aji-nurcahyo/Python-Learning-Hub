# Lab A02: Data Manipulation with Pandas
Wrangling dataset `retail_transactions_raw.csv`: normalisasi teks, deduplikasi, imputasi median kategori, dan agregasi omzet.

```python
from pathlib import Path
import pandas as pd

df = pd.read_csv(Path('datasets/retail_transactions_raw.csv')).copy()
df['city'] = df['city'].str.strip().str.title()
df['category'] = df['category'].str.strip().str.title()
df = df.drop_duplicates(subset=['order_id']).reset_index(drop=True)
df['quantity'] = df['quantity'].fillna(1.0)
df['unit_price'] = df['unit_price'].fillna(df.groupby('category')['unit_price'].transform('median'))
df['revenue'] = df['quantity'] * df['unit_price']
assert df['order_id'].is_unique
print(df.groupby('city')['revenue'].sum())
```
