# Lab B04: Model Evaluation, Cross-Validation & Tuning
Evaluasi `StratifiedKFold`, `GridSearchCV`, ROC-AUC, dan penyesuaian threshold pada `fraud_alert_sample.csv`.

```python
from pathlib import Path
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import roc_auc_score
from sklearn.model_selection import GridSearchCV, StratifiedKFold, train_test_split
from sklearn.pipeline import Pipeline

df = pd.read_csv(Path('datasets/fraud_alert_sample.csv'))
feats = ['amount_k', 'night_tx', 'velocity_1h', 'distance_km']
X_tr, X_te, y_tr, y_te = train_test_split(df[feats], df['fraud_flag'], test_size=0.25, stratify=df['fraud_flag'], random_state=42)
search = GridSearchCV(
    Pipeline([('rf', RandomForestClassifier(random_state=42))]),
    {'rf__n_estimators': [20, 50], 'rf__max_depth': [2, 4]},
    cv=StratifiedKFold(n_splits=4, shuffle=True, random_state=42),
    scoring='roc_auc'
).fit(X_tr, y_tr)
print('Holdout ROC-AUC:', round(float(roc_auc_score(y_te, search.predict_proba(X_te)[:, 1])), 4))
```
